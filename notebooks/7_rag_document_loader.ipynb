#### TextLoader from the text file

```python
from langchain_community.document_loaders import TextLoader
file_path = r"D:\TextLoaderExample.txt"
loader= TextLoader(file_path,encoding="utf-8")
text=loader.load()
print(text[0].page_content)
```
