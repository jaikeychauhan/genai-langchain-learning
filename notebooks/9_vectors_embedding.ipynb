```python
from dotenv import load_dotenv
load_dotenv()
```

```python
text=   [
    "I have done some python codes files in local in visual studio code.",
    "I have github account and want to create a new repository to checkin this"
]
```

```python
from langchain_openai import OpenAIEmbeddings
import numpy as np
embeddings=OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
vector=embeddings.embed_documents(text)
```

#### *********** STORE VECTOR EMBEDDINGS IN VECTOR DB ***************

```python
from langchain_chroma import Chroma
```

```python
vector_store=Chroma.from_texts(
    texts=text,
    embedding=embeddings,
    persist_directory="./vectorDB"   
)
```

```python
query="git hub"
result=vector_store.similarity_search(query=query,k=1)
print(result)
```
