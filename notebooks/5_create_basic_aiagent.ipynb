```python
from langchain_core.tools import tool
@tool
def add_numbers(a:int, b:int):
    """
        it will return the sum of two nubers
        Args:
            a: Number one
            b: Number two
    """
    return a+b

@tool
def multiply_numbers(a:int, b:int):
    """
        it will return the product of two nubers
        Args:
            a: Number one
            b: Number two
    """
    return a*b
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm=ChatOpenAI(model="gpt-4")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are math teacher, and always use tools for calculation."
)
```

```python
response=agent.invoke({"messages":[{"role":"user","content":"what is 2 + 3?"}]})
```
