```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")
search=GoogleSerperAPIWrapper()
#res=llm.invoke("who won the ICC cricket woman's world cup 2025?")
#print(res.content)
```

```python
agent=create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are a agent and you can search any question on google."
)
```

```python
#question="what is the address of techsimplus learning?"
question="which openai model recently launched?"
response=agent.invoke({"messages":[{"role":"user","content":question}]})
```

```python
print(response['messages'][-1].content)
```
